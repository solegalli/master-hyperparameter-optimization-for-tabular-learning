## Covariance

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(10)

n = 100

x1 = np.random.normal(loc=0.0, scale=1.0, size=n)
x2 = np.random.normal(loc=0.0, scale=1.0, size=n)

plt.scatter(x1,x2)
plt.ylim(-3, 3)
plt.xlim(-3, 3)
plt.axhline(y=0, color='black')
plt.axvline(x=0, color='black')
plt.xlabel('X1')
plt.ylabel('X2')
plt.savefig('covariance_uncorrelated_variables.png', dpi=320, bbox_inches='tight')
```

```python
np.sum( (x1 - np.mean(x1)) * (x2 - np.mean(x2))) / n
```

```python
np.cov(x1, x2)
```

```python
x1 = np.random.normal(loc=0.0, scale=1.0, size=n)
x2 = 2*x1 + np.random.normal(loc=0.0, scale=1.0, size=n)

plt.scatter(x1,x2)
plt.ylim(-8, 8)
plt.xlim(-3, 3)
plt.axhline(y=0, color='black')
plt.axvline(x=0, color='black')
plt.xlabel('X1')
plt.ylabel('X2')
plt.savefig('covariance_positive_relationship.png', dpi=320, bbox_inches='tight')
```

```python
np.sum( (x1 - np.mean(x1)) * (x2 - np.mean(x2))) / n
```

```python
np.cov(x1, x2)
```

```python
x1 = np.random.normal(loc=0.0, scale=1.0, size=n)
x2 = -2*x1 + np.random.normal(loc=0.0, scale=1.0, size=n)

plt.scatter(x1,x2)
plt.ylim(-8, 8)
plt.xlim(-3, 3)
plt.axhline(y=0, color='black')
plt.axvline(x=0, color='black')
plt.xlabel('X1')
plt.ylabel('X2')
plt.savefig('covariance_negative_relationship.png', dpi=320, bbox_inches='tight')
```

```python
np.sum( (x1 - np.mean(x1)) * (x2 - np.mean(x2))) / n
```

```python
np.cov(x1, x2)
```
