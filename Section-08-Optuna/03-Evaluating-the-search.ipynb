# Search Analysis with Optuna

Notebook for the course [Master Hyperparameter Optimization for Tabular Learning](http://www.trainindata.com/p/master-hyperparameter-optimization-for-tabular-learning)

Optuna has various built-in functions to examine the progress of the search.

In this notebook, I will demo the built-in plotting functions from Optuna.

We will use the [matplotlib backend](https://optuna.readthedocs.io/en/stable/reference/visualization/matplotlib/index.html#module-optuna.visualization.matplotlib). Optuna also offers the functions with [plotly backend](https://optuna.readthedocs.io/en/stable/reference/visualization/index.html).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import optuna
```

## Load the search results

We created and stored this search in the previous notebook.

```python
study_name = "lightgbm_study"  # produced by notebook 02
storage_name = "sqlite:///{}.db".format(study_name)

study = optuna.load_study(
    study_name=study_name,
    storage=storage_name,
)
```

```python
study.best_params
```

## Optimization history

```python
fig = optuna.visualization.matplotlib.plot_optimization_history(study)
```

## Contour plot - 2D plots

```python
optuna.visualization.matplotlib.plot_contour(
    study,
    params=["num_leaves", "max_depth", "learning_rate", "num_boost_round"],
)
plt.show()
```

## Plot slice

```python
optuna.visualization.matplotlib.plot_slice(
    study,
    params=["num_leaves", "max_depth", "learning_rate", "num_boost_round"],
)
plt.show()
```

## Parameter importance

fAnova by default but there are other methods. More details [here](https://optuna.readthedocs.io/en/stable/reference/importance.html)

```python
optuna.visualization.matplotlib.plot_param_importances(study)
plt.show()
```

## Plot parallel coordinate

Plots the high dimensional space of the search. In plain English, the feature combinations.

```python
optuna.visualization.matplotlib.plot_parallel_coordinate(   
    study,
    params=["num_leaves", "max_depth", "learning_rate", "num_boost_round"],
)
plt.show()
```
